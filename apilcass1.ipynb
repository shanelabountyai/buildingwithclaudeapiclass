```python
import subprocess
import os
from pathlib import Path

# Read the key directly from .env file
project_root = Path("/Users/shanelabounty/python/buildingwithclaudeapiclass/my-python-project")
env_file = project_root / ".env"

# Parse .env manually
api_key = None
if env_file.exists():
    with open(env_file) as f:
        for line in f:
            if line.startswith("ANTHROPIC_API_KEY="):
                api_key = line.split("=", 1)[1].strip()
                break

if not api_key:
    print("ERROR: Could not read API key from .env")
else:
    print(f"API key loaded from file: {bool(api_key)} (length: {len(api_key)})")
    
    # Run the API call in the venv with the key passed explicitly
    venv_python = project_root / ".venv/bin/python"
    
    code = f"""
import os
os.environ["ANTHROPIC_API_KEY"] = "{api_key}"

from anthropic import Anthropic

client = Anthropic()
response = client.messages.create(
    model="claude-haiku-4-5-20251001",
    max_tokens=200,
    messages=[{{"role": "user", "content": "What is quantum computing in two sentences?"}}]
)
print(response.content[0].text)
"""
    
    result = subprocess.run(
        [str(venv_python), "-c", code],
        cwd=project_root,
        capture_output=True,
        text=True
    )
    
    if result.returncode != 0:
        print("Error:")
        print(result.stderr)
    else:
        print(result.stdout)
```
